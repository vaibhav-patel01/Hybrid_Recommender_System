```python
import pandas as pd 
import ast
```

```python
df = pd.read_csv("cleaned_metadata.csv", converters={
    'genres': ast.literal_eval,
    'studios': ast.literal_eval, 
    "title": ast.literal_eval,
}, keep_default_na=False)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)
```

```python
urls = pd.read_csv("metadata")
```

```python
df.head(5)
```

```python
urls = urls[["id", "coverImage", "bannerImage", "trailer"]]
```

```python
urls['coverImage'] = urls['coverImage'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) and x != '' else "coverImage unavailable"
)
urls['trailer'] = urls['trailer'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) and x != '' else "trailer unavailable"
)
```

```python

# urls["bannerImage"] = urls["bannerImage"].fillna("bannerImage unavailable")
urls["trailer"] = urls["trailer"].fillna("trailer unavailable")
```

```python
urls['coverImage'] = urls['coverImage'].apply(
    lambda x: x.get('large') if isinstance(x, dict) else None
)
```

```python
urls['trailer'] = urls['trailer'].apply(
    lambda x: x.get('id') if isinstance(x, dict) else None
)
```

```python
urls['trailer'] = urls['trailer'].apply(
    lambda x: f"https://www.youtube.com/watch?v={x}" if pd.notna(x) else None
)
```
